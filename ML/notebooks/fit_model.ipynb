```python
def get_two_weeks_before(date: datetime) -> datetime:
    """
    Returns a datetime object that is 2 weeks (14 days) before the input date
    
    Args:
        date (datetime): The reference date
        
    Returns:
        datetime: A datetime object 14 days before the input date
    """
    return date - pd.Timedelta(days=14)
```

```python
from typing import List, Dict, Tuple
import pandas as pd
import numpy as np
from darts import TimeSeries
from darts.models import LightGBMModel
from darts.utils.timeseries_generation import datetime_attribute_timeseries


def polars_to_pandas(df_pl: pl.DataFrame) -> pd.DataFrame:
    return df_pl.to_pandas()


def build_timeseries_for_darts(
    df_pl: pl.DataFrame,
    route_id_col: str = "route_id",
    stop_id_col: str = "stop_id",
    datetime_col: str = "current_datetime",
    target_col: str = "last_arrival_delay",
    covariate_cols: List[str] = None,
) -> Tuple[List[TimeSeries], List[TimeSeries], List[Dict]]:
    """
    Build lists of (target_series, covariate_series) for each (route_id, stop_id).
    """
    if covariate_cols is None:
        covariate_cols = []

    pdf = polars_to_pandas(df_pl).copy()
    pdf.sort_values(datetime_col, inplace=True)

    # We'll build multiple time series, one per route-stop
    target_series_list = []
    covariate_series_list = []
    series_metadata = []

    combos = pdf[[route_id_col, stop_id_col]].drop_duplicates()
    for _, row in combos.iterrows():
        rid = row[route_id_col]
        sid = row[stop_id_col]

        subset = pdf[(pdf[route_id_col] == rid) & (pdf[stop_id_col] == sid)].copy()
        if subset.empty:
            continue

        # Darts TimeSeries
        ts_target = TimeSeries.from_dataframe(
            df=subset,
            time_col=datetime_col,
            value_cols=target_col,
            fill_missing_dates=True,  # fill with NaNs
            freq="infer",  # or specify if known
        )

        # Build covariates
        if covariate_cols:
            ts_cov = TimeSeries.from_dataframe(
                df=subset,
                time_col=datetime_col,
                value_cols=covariate_cols,
                fill_missing_dates=True,
                freq="infer",
            )
        else:
            ts_cov = None

        target_series_list.append(ts_target)
        covariate_series_list.append(ts_cov)
        series_metadata.append({"route_id": rid, "stop_id": sid})

    return target_series_list, covariate_series_list, series_metadata


def train_lightgbm_global(
    target_series: List[TimeSeries],
    covariate_series: List[TimeSeries],
    lags: int = 12,
    lags_past_covariates: int = 12,
) -> LightGBMModel:
    """
    Train a global LightGBM model using Darts.
    """
    model = LightGBMModel(
        lags=lags, lags_past_covariates=lags_past_covariates, random_state=42
    )
    model.fit(target_series, past_covariates=covariate_series)
    return model
```
