```python
import polars as pl
from datetime import datetime
import pandas as pd
import h3

pd.set_option("display.max_columns", None)
import altair as alt

alt.data_transformers.enable("vegafusion")
```

```python
translink_df = pl.scan_parquet("../data/processed/merged_translink_df_full.parquet")
```

```python
translink_df.head().collect().to_pandas().to_dict('records')
```

```python
def lat_lon_to_h3(lat, lon, resolution):
    return h3.geo_to_h3(lat, lon, resolution)
```

```python
translink_df.with_columns(
    pl.struct(["vehicle_latitude", "vehicle_longitude"])
    .map_elements(
        lambda struct: lat_lon_to_h3(
            struct["vehicle_latitude"], struct["vehicle_longitude"], 7
        )
    )
    .alias("vehicle_h3_7")
).head().collect()
```
